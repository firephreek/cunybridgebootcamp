# Python 2 vs Python 3 (Less important nowadays but worth mentioning!)

For most programming languages, it’s safe to assume everyone is using the latest stable version. Python was an exception for a long time:

- Python 2.7 was released in 2010.

- Python 3 first appeared in 2008, but adoption was slow because it was not backward-compatible — Python 2 code often had to be rewritten to work in Python 3.

This split meant the astronomy community (and many other scientific fields) lived in both worlds for years.
However:

- Python 2 reached end-of-life on January 1, 2020. It no longer receives updates, including security fixes.

- Modern scientific Python libraries — including numpy, astropy, and pandas — are Python 3 only.

- You’ll only encounter Python 2 today when dealing with very old, unmaintained software or legacy data pipelines.

If you do stumble across Python 2 code, it may break in Python 3 due to a few key differences:
There's a few basic differences to mention below, but more can be found [here](http://sebastianraschka.com/Articles/2014_python_2_3_key_diff.html), with examples of how the code runs in both Python 2 and Python 3.


# Print statements
The second big difference is that there's additional formatting required for print statements. This will mean that Python 2 code with any print statements in the old format will have a syntax error when it's run in Python 3. A quick example:

```
print 'Hello World!' #the Python 2 way
```

```
print('Hello World!') #the Python 3 way
```

## Integer division:
In Python 2, when dividing integers, the result was always an integer. In Python3, the result always includes decimals, making integer division more intuitive.

Python 2:
```
>>> 5/2
2
```

Python 3:
```
>>> 5/2
2.5
```

## Unicode support:
In Python 2, every single Unicode string has to be marked with a “u” prefix as it uses ASCII characters by default. By contrast, Python 3 stores strings as Unicode by default, which is more versatile than ASCII strings. 

Python 2:
```
>>> print type("Hello DataCamp!") # this is an ASCII string
<type 'str'>
```
Python 3:
```
>>> print type(u"Hello DataCamp!") # this is a Unicode string
<type 'unicode'>
```

## Range function:

Range function
The `xrange()` function in Python 2 no longer exists in Python 3. It has been replaced by the `range()` function, which has improved performance when iterating over sequences

Python 2:
```
>>> xrange(1,10)
xrange(1, 10)
>>> type(xrange(1,10))
<type 'xrange'>
```

Python 3:
```
>>> range(1,10)
range(1, 10)
>>> type(range(1,10))
<class 'range'>
```


# Some resources for converting old Python 2 code to Python 3 
#### (Warning! I haven't tested these...)

Fortunately, there are some resources that can help convert Python 2 code to Python3.

[2to3](https://docs.python.org/3/library/2to3.html): A Python program that takes Python 2.x source code and applies a series of fixers to transform it into valid Python 3.x code.

[Python-Future](https://python-future.org/#:~:text=python%2Dfuture%20is%20the%20missing,Python%203%20with%20minimal%20overhead.): Allows you to use a single, clean Python 3.x-compatible codebase to support both Python 2 and Python 3 with minimal overhead.

[Six](https://pypi.org/project/six/): A Python 2 and 3 compatibility library. It provides utility functions for smoothing over the differences between the Python versions with the goal of writing Python code that is compatible with both Python versions.
